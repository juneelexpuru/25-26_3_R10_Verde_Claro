## 3. Obtención de los looks

### Triángulos

```python
num_triangulos = nx.triangles(G)
sum(num_triangulos.values()) // 3
```

```python
G_igraph = Graph.from_networkx(G)
triangulos = G_igraph.cliques(min=3, max=3)
```

### Filtrar los triangulos para que cumplan las reglas 9, 10 y 11.

**Regla 9** : al generar tríos (looks), al menos una de las prendas conectadas debe tener basic = True.

```python
basic_nodes = set(df_grafo.index[df_grafo["basic"]])

triangulos_filtrados_basic = [
    t for t in triangulos
    if any(n in basic_nodes for n in t)
]
```

**Regla 10 y 11**: los looks tienen que tener siempre una prenda de nivel 1 o 2. Si el look contiene una prenda primer nivel, tiene que tener dos de cuarto nivel. Si el look contiene una prenda de segundo nivel, tiene que tener uno de tercer y otro de cuarto nivel.

```python
nivel_dict = df_grafo["nivel"].to_dict()
triangulos_filtrados = filtrar_triangulos(triangulos_filtrados_basic, nivel_dict)
```

### Función para obtener los looks

Se han generado una cantidad de 15 looks

```python
def generar_looks_10(df_expandido, triangulos_filtrados):
    import pandas as pd

    # --- 0) Formatos ---
    df = df_expandido.copy()

    # Si ya cambiaste el nombre afuera, esto asegura que la función lo encuentre
    if "index" not in df.columns:
        if "indice" in df.columns:
            df = df.rename(columns={'indice': 'index'})
        else:
            raise ValueError("df_expandido debe tener una columna 'index' o 'indice'.")

    df["index"] = df["index"].astype(int)
    df["hexadecimal"] = (
        df["hexadecimal"].astype(str).str.strip().str.lstrip("#").str.upper()
    )

    # --- 1) Acceso rápido ---
    row_by_node = df.set_index("index", drop=True)

    # --- 2) Detectores ---
    def _col(df_look, name):
        if name not in df_look.columns:
            return pd.Series([""] * len(df_look), index=df_look.index)
        return df_look[name]

    def tiene_vestido(df_look):
        cat = _col(df_look, "categoria_prenda").astype(str)
        sub = _col(df_look, "subcategoria_prenda").astype(str)
        return cat.str.lower().isin({"dress"}).any() or sub.str.contains("dress", case=False, na=False).any()

    def tiene_falda(df_look):
        sub = _col(df_look, "subcategoria_prenda").astype(str)
        return sub.str.contains("skirt", case=False, na=False).any()

    def tiene_pantalon(df_look):
        sub = _col(df_look, "subcategoria_prenda").astype(str)
        return sub.str.contains("pant|trouser|jean|cigarette", case=False, na=False).any()

    def tiene_accesorio(df_look):
        cat = _col(df_look, "categoria_prenda").astype(str)
        sub = _col(df_look, "subcategoria_prenda").astype(str)
        niv = _col(df_look, "nivel")
        try:
            niv_num = pd.to_numeric(niv, errors="coerce")
        except Exception:
            niv_num = pd.Series([None] * len(df_look), index=df_look.index)
        return (
            cat.str.lower().isin({"accessories"}).any()
            or sub.str.contains("bag|foulard|scarf|belt", case=False, na=False).any()
            or (niv_num == 4).any()
        )

    # --- 3) Análisis ---
    HEX_AZULES = {"00008B", "000080", "153668", "1E90FF", "0B3D91"}

    def analiza_look(nodes):
        # Al llegar aquí ya sabemos que los nodos existen gracias al check en 'pasada'
        df_look = row_by_node.loc[list(nodes)]
        colores = set(_col(df_look, "hexadecimal").unique())
        color_distinto = len(colores) > 1

        return {
            "df": df_look,
            "colores": colores,
            "color_distinto": color_distinto,
            "azul": len(colores & HEX_AZULES) > 0,
            "vestido": tiene_vestido(df_look),
            "falda": tiene_falda(df_look),
            "pantalon": tiene_pantalon(df_look),
            "accesorio": tiene_accesorio(df_look),
        }

    # --- 4) Lógica de Selección ---
    target = {"vestidos": 2, "faldas": 2, "pantalones": 3, "accesorios": 3}
    PRIORIDAD = ["vestidos", "faldas", "pantalones", "accesorios"]
    looks = {"vestidos": [], "faldas": [], "pantalones": [], "accesorios": []}
    numeros_usados = set()
    blue_usado = {"faldas": False, "pantalones": False, "accesorios": False}

    def total_actual(): return sum(len(v) for v in looks.values())
    def cupo_lleno(cat): return len(looks[cat]) >= target[cat]
    
    def cumple_cat(cat, info):
        return (
            (cat == "vestidos" and info["vestido"]) or
            (cat == "faldas" and info["falda"]) or
            (cat == "pantalones" and info["pantalon"]) or
            (cat == "accesorios" and info["accesorio"])
        )

    def puede_entrar(cat, info):
        if cupo_lleno(cat): return False
        if not cumple_cat(cat, info): return False
        if cat in blue_usado and info["azul"] and blue_usado[cat]: return False
        return True

    def add_to(cat, look, info):
        looks[cat].append(look)
        numeros_usados.update(look)
        if cat in blue_usado and info["azul"]:
            blue_usado[cat] = True

    def pasada(exigir_color_distinto=True):
        for look in triangulos_filtrados:
            if total_actual() >= 10: break
            
            # --- CORRECCIÓN AQUÍ: Validar que todos los nodos del triángulo están en el DF ---
            if not all(n in row_by_node.index for n in look):
                continue
            
            if any(n in numeros_usados for n in look):
                continue

            info = analiza_look(look)
            if exigir_color_distinto and not info["color_distinto"]:
                continue

            for cat in PRIORIDAD:
                if puede_entrar(cat, info):
                    add_to(cat, look, info)
                    break

    pasada(exigir_color_distinto=True)
    if total_actual() < 10:
        pasada(exigir_color_distinto=False)

    return looks, row_by_node
```

```python
df_expandido = pd.read_csv("../Datos/Transformados/df_expandido.csv",sep=';')
```

```python
df_expandido = df_expandido.rename(columns={'indice': 'index'})
```

```python
looks, row_by_node = generar_looks_10(df_expandido, triangulos_filtrados)

print({k: len(v) for k, v in looks.items()}, "TOTAL =", sum(len(v) for v in looks.values()))
```
